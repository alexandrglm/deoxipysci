### Variables

- Una variable es "un espacio en la memoria", al que le otorgamos (asignamos) "un nombre" y un dato (un valor), y que "usamos" (a la que accedemos) posteriormente.

```python
variable = "Soy el dato de una variable"

nombre = "Pepe"
edad = 37
sabe_ingles = True
nota_de_corte = 7.75
```

- En Python, la convención para variables es la `snake_case` (palabras en minúsuclas, separadas por barra baja `_`)


### Tipos (de datos)

- Python es un lenguaje de tipado débil (para definir o referi el tipo de dato de una variable no tenemos que espeficar el tipo constantemente, como puede ocurrir con otros lenguajes).

- Además, Python es **zero-based**, es decir, los tipos numéricos **siempre se empieza a contar desde CERO**.
  
***

Los tipos básicos para Python son:  
  

*   Numéricos -> Enteros (Integers)
*   Numéricos -> Decimales (Floats)

*   Lógica binaria -> Booleanos:
    - `True`, que numéricamente equivale a '1'
    - `False`, que numéricamente equivale a '0'

*   Cadenas/Caracteres (Strings)

*   Estructuras de Datos:
    -   Listas -> `[]`
    -   Tuplas -> `()`
    -   Diccionarios -> `{}`


- Para conocer qué tipo de dato contiene una variable, usamos el la función `type(variable)`, la que nos devuelve la propiedad `type` de dicho valor.
