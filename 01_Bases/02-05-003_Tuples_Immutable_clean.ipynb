## TUPLAS `Tuple`

Una tupla, al igual que una lista, es un contenedor, una secuencia ordenada de elementos, **pero con sus propias características**:  

  - Se define con **paréntesis**  (parents, parenthesis) `()`.  
  
  - Los elementos siguen separándose por comas, y su acceso por index es exáctamente igual que en las listas.  
  - También tiene propiedades `type` y `length`

  - Es decir, se accede igual que con las listas ...
  
  - **PERO, la diferencia con la lista es ... en COMO (NO) SE MODIFICA**, porque no se puede, dada la característica de **IMMUTABILIDAD**.  

---

> **¿Por qué importa la immutabilidad de Tuplas?**
>  
>  Las tuplas **garantizan integridad**
>
> *Coordenadas fijas, hiperparámetros de un algoritmo, configuración que no debe cambiar en tiempo de ejecución...* se guardan en tuplas para que ninguna parte del código pueda modificarlas por accidente.  
>
> Son el equivalente a las constantes en otros lenguajes.

***

### IMMUTABILIDAD ABSOLUTA de `Tuple`

Una tupla es immutable:  
  * **Una vez creada NO se puede modificar**:
    - Ni añadir un elemento  
    - Ni reemplazar un elemento    
    - Ni eliminar un elemento  

Es decir, **tiene inmutabilidad de los valores que contienen**.
  
  * Intentar hacer lo contrario hará que Python te salude con un `AtributeError` o `TypeError`.  


```python
tupla = (1, 2, 3)

t.append(4) # AttributeError: 'tuple' object has no attribute 'append'

del t[0]    # TypeError: 'tuple' object doesn't support item deletion
```


Pero sí que hay un matiz sobre su immutabilidad, y tiene un "secreto", donde una tupla original sí puede ser modificada (donde **puedes hacer crecer dicha tupla**):
  
  * No podrás cambiar los valores de una tupla ya creada ...
  
  * ... **pero sí concatenar nuevos valores o tuplas**:

    1. Bien creando una nueva tupla  
    2. Bien "haciendo trampa", y reasignando a la misma variable la concatenación de dicha tupla consigo mismo o con otras.  


```python
a = (1, 2, 3)
b = (4, 5, 6)

# 1. NUEVA
c = a + b
print(c)    # (1, 2, 3, 4, 5, 6)
print(a)    # (1, 2, 3)
print(b)    # (4, 5, 6)

# 2. REASIGNANDO
a = a + b
print(a)   # (1, 2, 3, 4, 5, 6)

```
    
Es decir, una tupla **tiene siempre immutabilidad de los valores que contiene ... y reasignar nuevos valores a la misma tupla no tumba la immbutabilidad**:  

  * **No se ha modificado la tupla original** `(1, 2, 3)`. Sigue existiendo en memoria, intacta.  

  * `a + b` **crea una tupla nueva** `(1, 2, 3, 4, 5, 6)`.  
  
  * El `=` **reasigna el nombre** `a` para que apunte a esa tupla nueva.  
  
  * La tupla original `(1, 2, 3)` se queda sin nombre que la referencie y, en algún momento, Python la borrará de memoria.  


Todo ello se puede comprobar con la función legacy `id()`:

```python
a = (1, 2, 3)
print(id(a))     # Ej: 140156066514016

a = a + (4, 5, 6)
print(id(a))     # Ej: 140156067649504
```

***


### OPERACIONES PERMITIDAS

Aunque sea inmutable, una tupla se puede **leer y combinar** como una lista:

  - **Indexación**
    * `t[0]`, `t[-1]`

  - **Slicing**
    * `t[1:3]`, `t[::-1]`, etc. (devuelve una tupla nueva)

  - **`len()`**
    * Cuenta elementos

  - **Concatenación con `+`**
    - Crea una tupla nueva con los elementos de ambas

  - **Replicación con `*`**
    * `t * 3` repite la tupla
  
---
Como hemos indicado antes, lo que **no** se puede:  
    
  * Ni asignar por index
  
  * Ni métodos `del`, `.append()`, `.remove()`, etc.  
  
  * NADA que modifique la estructura de una tupla.    
---

### Empaquetado y desempaquetado (packing / unpacking)

Aquí está la parte que más confunde al principio, y es la más útil en la práctica.

- **Empaquetado (packing)**:  *si asignas varios valores separados por comas a **una sola variable**, Python crea automáticamente una tupla*

```python
coords = 10, 20        # sin paréntesis, CREA UNA TUPLA (10, 20)

coords = (10, 20)      # con paréntesis, ES LO MISMO
```

> *Los paréntesis son opcionales; lo que define la tupla es la **coma***.
  

- **Desempaquetado (unpacking)**:   *si asignas una tupla a **varias variables** a la vez, Python reparte los valores en orden*:
  * El número de variables a la izquierda debe coincidir con el número de elementos de la tupla, o te saluda un `ValueError`.  

```python
coords = (10, 20)

x, y = coords  # x = 10, y = 20
```


-  **Varias variables para varios valores** (unpacking) ≠ **una variable para varios valores** (packing, crea tupla).
-  Ambas se escriben con comas; lo que cambia es el lado izquierdo del `=`.
