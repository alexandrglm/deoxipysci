### FUNCIONES

Una función es **un bloque de código encapsulado, y con nombre propio, que se define una vez y puede ser reutilizada** llamándola por su nombre.  

1.   Recibe datos
2.   Realiza una lógica con ellos
3.   Devuelve un resultado


***



#### Funciones vs Métodos

Ambos son bloques de código reutilizables, pero se invocan de forma distinta:


1.   **FUNCIÓN**  
      *   Independiente.  
      *   Se llama por su nommbre (`print()`, `len()`, `range()`)  

2.   **MÉTODO**  
      *   Pertenece a una variable
      *   Se llama con **notación de punto** (`lista.append()`, `matriz.copy()`, `string.upper()`)

> **Una función vive "suelta" en el código, y un método vive "pegado" a un objeto.**


***




###  **Def**inir una función `def:`

Sintácticamente, una función tiene estos componentes:  

  *   `def`  
  *   Nombre de la función  
  *   `()` Paréntesis, incluye **parámetros** (si los necesitase)  
  *    `:` semicolon  
  *   Bloque indentado  
  *   `return`, generalmente  


```python
def mi_funcion(x, y):

  resultado = x + y

  return resultado
```

Definir una función no la ejecuta, y hay que invocarla posteriormente:  

```python
mi_funcion(2, 3)
mi_funcion(24, 12)
mi_funcion(8, 2)
```
***


#### Parámetros `(x, y)`

Los parámetros son las variables internas que la función recibe y con las que la función va a operar.  

Sintácticamente, van dentro de los paréntesis, y son el motivo por el que una función es reutilizable.  

En el ejemplo del bloque anterior, parámetros serían cada factor que después se emplea para hacer la suma.  

Existen dos formas de pasar argumentos:  

1. **POSICIONAL**
    * El el mismo orden en que se definen  
    * `suma(3, 5)`... `x=3` e `y=5`

2. **POR NOMBRE** (keyword)
    * Expresando explícitamente el nombre del parámetro y su valor  
    * `suma(x=3, y=5)`, mismo resultado  
    * Cuando una función tiene muchos parámetros, el paso por nombre es la práctica recomendable  


Si una función tiene dos parámetros y solo pasas uno, te saludará un `TypeError`:
> **El número de argumentos debe coincidir con el número de parámetros en la definición** (aunque hay ciertas trampas como usar `default` en alguno de ellos).

***

### `return`

`return` es lo que devuelve la función a la parte del código que la invocó.  

Sin `returnp`, la función hace ... su función, pero no entrega nada útil (devuelve `None`).  

Diferencia para entenderlo:  

  * `print(...`) dentro de una función:
    - Muestra algo en pantalla, pero no entrega nada al código llamante.  
    - Una función con solo print devuelve None.

  * `return ...`:
    - Entrega el valor a quien llamó.  
    - Permite guardarlo en una variable, pasarlo a otra función, imprimirlo después, etc...

```python
def suma_print(x, y):

  print(x + y)


def suma_return(x, y):

  return(x + y)


a = suma_print(3, 2)
b = suma_return(3, 4)

print(a)
print(b)
```
  
---

> 💡 Como apunte, en IA y Ciencia de datos las funciones casi nunca usan `print()` (no necesitan mostrar en consola cosas); devuelven datos al pipeline para ser procesados, por lo que `return` es lo que hace que una función sea componible con el resto del código.  

***


#### Scope (Ámbito):  `local` vs `global`

  
Cada variable vive dentro de un tipo de ámbito u otro:

  *   **LOCAL SCOPE**, ámbito local, variables creadas dentro de una función o un bloque de lógica concretos. Solo existen mientras se ejecutan. Al terminar, mueren.  

  *   **GLOBAL SCOPE**, ámbito global, creadas en/para el programa principal, no específicas de una función o bloque de lógica. Viven durante toda la ejecución.  

Una función **puede leer variables locales y globales** sin problema, pero **no puede modificar variables globales sin permiso explícito**:

  * Si dentro de una función intentas modificar una variable global, Python crea una variable local con el mismo nombre, sin tocar la global.  
  * Para modificar la variable global de verdad, debe declararse con la palabra `global`


```python
counter = 0

def incremental():

  global counter
  counter +=1


print(counter) # Dará 0
incremental()
print(counter) # Dará 1
incremental()
incremental()
print(counter) # Dará 3
```
  


> **REGLA MENTAL:**    
>   
> `global` se usa con mucha moderación. Sirve para casos concretos (como contadores, configuraciones mutables de alcance completo, ...).  
>   
> Para todo lo demás, lo correcto es pasar datos por parámetros y devolver resultados con `return`
