## Control Flow Statements

Hasta ahora, hemos visto código estático:  


*   Se ejecutaba linea por linea, de arriba abajo
*   Sin nada que lo dirigiera, sin desviarse según una situación u otra

En un programa real, el código necesita **tomar decisiones**, o **repetir tareas**.  

Para esto está el control de flujo, con lo que un programa deja de ser "una simple calculadora", y pasa a ser una herramienta, un programa.  

En el control de flujo hay **DOS herramientas** fundamentales:  

*   **RAMIFICACIÓN** (`if - elif - else`)
    *   Elegir qué bloque se ejecuta, según condiciones

*   **ITERACIÓN** (`while`, `for`)
    *   Repetir un bloque según condiciones

Ambas dependen de booleanos y de operadores lógicos.  

***




### RAMIFICACIÓN `if` `elif` `else`

Estructura básica:  

```python
if condicion_1:
  
  # Bloque que se ejecuta si la condición 1 es True

elif condicion_2:

  # Se ejecuta si la 1 no aplica y fue False y condición 2 es True

else:

  # Si todo lo anterior es False, salvaguarda aquí

```
  
Como todo, el control de flujo tiene sus reglas:  

*   `if` es obligatorio, `elif` y `else`son opcionales
*   Sintácticamente, siempre termina cada expresión de ramificación con un semicolon `:`
*   La indentación sigue siendo crucial y obligatoria
*   El bloque entero de ramificación se ejecuta de arriba abajo y, en cuanto encuentra un `True` se salta todo el bloque, por lo que los `elif` posteriores no comprueban nada.
*    Si ninguna condición es `True`, se pasa al `else`; si este no existe, no se ejecuta nada y se continúa con el código posterior.  
* Dentro de cada condición se pueden combinar operadores lógicos `and`, `or`, `not`.  


```python
termostato = 16

if termostato > 25:
  print("Demasiado calor, se está desperdiciando energía")

elif termostato >= 21 and termostato < 24:
  print("Calor, es el momento de apagar la caldera")

elif termostato >= 18 and termostato < 21:
  print("termostato óptima")

else:
  print("Frío, es hora de encender la caldera")

```

***

### ORDEN DE ESPECIFICAD (Trampa de prioridad)

Como se evalúa de arriba abajo y se detiene ante la primera condición `True`, **el orden importa**.  

Si al comienzo usas una regla "muy amplia", ninguna de las posibles condiciones posteriores tendrán uso.  



> **PRIMERO DE LO MÁS ESPECÍFICO, DESPUÉS LO MÁS GENERAL**
*   Primero las salvaguardas
*   Primero las condiciones previas
*   Primero los casos que queremos filtrar
*   Primero los casos específico
*   ...
*   Una vez tenemos todo lo específico concretado, aplicamos lo más general

---

- ⛔ Ejemplo incorrecto (nunca llega a `elif` porque cualquier temperatura superior a 0º ya encaja con la primera condición):  
```python
if termostato > 0:
  print("Temperatura positiva")

elif temperatura == 20:
  print("Temperatura ideal")
```

- 👌Ejemplo correcto
```python
if temperatura >= 20:
  print("Temperatura ideal")

elif termostato > 0:
  print("Temperatura positiva")

else:
  print("Temperatura negativa")

```

### ITERACIÓN `while`

`while` repite un bloque mientras la condición sea `True`:  
*   Se usa cuando **no sabes cuántas iteraciones habrás, pero sí cuándo parar**.
  
```python
x = 0

while x < 4:
  
  print(x)
  x += 1
```

Los blucles también tienen su propias reglas:  


1.   Una variable de control inicializada antes del bucle (`x = 0`)
2.   La condición del bucle `while` (`while x < 4:)`, siempre con semicolon `:`
3.   Una actualización de la variable de control dentro del bucle (`x += 1`)  
  

> ❌ **Si no se actualiza la variable de control, la condición nunca se volverá `False` y el bucle se ejecutará sin parar:  Bucle Infinito** ❌
>   
>    * Cada bucle debe contener "su propia autodestrucción".  
>    * Si no se "auto-anula", el bucle es infinito.  


***







### ITERACIÓN `for` `in` / `range()`

`for` recorre una estructura (listas, tuplas, strings, ...) **tomando un elemento por cada iteración**:  
  * **De antemano ya sabes cuántas iteraciones habrá**
    * Tantas como elementos tenga la estructura

```python
animals = ["dog", "cat", "bird"]

for animal in animals:

  print(animal)

```
* El bloque indentado se ejecuta una vez por cada elemento de la estructura.  
*   `animal` es una variable interna y temporal, que puedes llamarla como consideres.  
    *   En la primera iteración será `dog`, luego `cat` y finalmente `bird`.  

---
  
`range(inicio, fin, paso)` es una función legacy, **genera dinámicamente secuencias de números al vuelo**:  
  
  * No hace falta escribir una lista de números a mano

Tiene sus propias reglas también:  

  * `inicio`, por defecto es `0`
  * `fin`, como en los métodos de indexación, el parámetro fin **es autoexcluyente**
  * `paso`, también existe aquí, por defecto es `1`, y permite mostrar "de tanto en tanto" se le indique
  *  Paso en negativo, en cambio, aplica la inversa


```python
range(4)
range(2, 6)
range(0, 11, 2)
range(0, 11, -1)

for i in range(1, 101, 10):
  print(i)
```
