## DICCIONARIOS `{}`

Un diccionario, `dict` es otro tipo de contenedor, pero con características muy especiales:

   1. Es una **ESTRUCTURA DE DATOS BIDIMENSIONAL**:  
      - Ni guarda valores sueltos.  
      - Ni se indexa por posición numérica.  

   2. Almacena sus valores por **CLAVE-VALOR**, `{key: value}`:  

      * **CLAVE - KEY** es una etiqueta única que identifica el dato.  
      * **VALOR - VALUE** es el contenido, el dato, asociado a la clave.  

   3. **No es inmutable**, puedes modificar sus pares sin problema.  

  
   4. Sintácticamente:

    - Se define con llaves (braces, curly braces) `{ }`  
    - Cada `key` se une a su respectivo `value` con semicolon `:`  
    - Cada par `key: value` va precedido por coma `,`  



```python
dict = {}

dict = {
  "name": "Pepe",
  "age": 32,
  "location": "Antarctica"
}

```

---

**El diccionario es la representación en código de una fila de un dataset**:    

  - **Cada columna es una clave**.  

  - **Cada fila, cada celda, es un valor.**  


---

> **¿POR QUÉ ES IMPORTANT PARA ML?**  
>  
> Los datasets reales tienen Missing Data (faltan claves en muchas filas).  
>   
> Si tu modelo busca `"temperature"` con corchetes y falta en una fila de un millón, todo el entrenamiento colapsa por un `KeyError`.  
>  
>  Acostumbrarse a `.keys()` para inspeccionar y `.get()` con valores neutros (`0`, `-1`, `None`) es lo que separa un script frágil de un pipeline solvente para producción.  

---

***

### ACCESO POR CLAVE y `KeyError`


Una forma para acceder a un valor, de una clave determinada, es:  

  * Usando corchetes [ ], como accediendo a listas ...  
  * ... pero usando el nombre de la clave.

    - `dict["nombre de la clave]`  

Usar claves que no existen, hará que Python nos salude con un `KeyError`.

```python
client = {"name": "Ana Lopez", "age": 28}

print(client["name"])
print(client["age"])


print(client["inexistant key"])
```


***


### INSPECCIONANDO UN `dict` DESCONOCIDO

**Cuando cargas un dataset del que no sabes nada, primero lo auditas**.  

Tres métodos nativos, propios de `dict` permiten diseccionar el diccionario:  

- **`.keys()`** → Todas las claves.

  ```python
 client = {"name": "Ana", "age": 28}
  
  print(client.keys())
  # dict_keys(['name', 'age'])
  ```

- **`.values()`** → Todos los valores.

  ```python
  print(client.values())
  # dict_values(['Ana', 28])
  ```

- **`.items()`** → Todos los pares clave-valor, como pares (tuplas de dos elementos).

  ```python
  print(client.items())
  # dict_items([('name', 'Ana'), ('age', 28)])
  ```


Lo que devuelven estos métodos (`dict_keys`, `dict_values`, `dict_items`) slo have en **modo vista** (view) (no crea nuevos objetos en memoria):

  * No son listas, pero se pueden recorrer y convertir a lista con `list(...)` de ser necesario (a este nivel basta con saber que existen y qué contienen).

***

### VALIDACION de CLAVE con `in`

Esta es la primera línea de defensa contra el `KeyError`: *pregunto, luego pido*

Antes de pedir un valor, puedes **comprobar si la clave existe con el operador `in`, que devuelve un boolean**:

```python
client = {"name": "Ana", "age": 28}


print("name" in client)     # True
print("email" in client)    # False
```

`in`, sobre un diccionario, **comprueba claves, no valores**:

  - `"Ana" in client` devolvería `False` porque `"Ana"` es un valor, no una clave.  

***


### EXTRACCIÓN SEGURA con `.get()`

1. `.get(clave)` hace lo mismo que `person[clave]`, **PERO**:
    
    * **NO lanza `KeyError`** si la clave no existe.  

    * Devuelve `None` en su lugar.  


```python
client = {"name": "Ana", "age": 28}


print(client.get("name"))   # "Ana"
print(client.get("email"))  # None
```

---

2. `.get(clave, default)` añade un **valor por defecto** que se devuelve en lugar de `None` si la clave no existe:

    *  El valor por defecto sólo ocurre en View, **no se guarda** en el diccionario.  

    * Si la clave **SÍ existe**, `.get()` devuelve su valor real y **el default se desecha**.  

    * **Es la salvaguarda para operar con diccionarios ante valores null, vacíos o que no te esperas sin detener toda la ejecución**  

```python
print(client["email"])    # keyError
print("email" in client)  # False

print(clien.get("email")) # None (literalmnente nada en terminal)
print(client.get("email", "no disponible"))   # "no disponible"
print(client.get("name", "desconocido"))   # "Ana", el default se ignora
```

***


### NIVELES DE SEGURIDAD (o NO) EN LA EXTRACCIÓN

1. **Nivel básico (peligroso)**

    * `person["email"]` sin comprobar.  
    
    * Si la clave falta, `KeyError` y crash.  


2. **Nivel intermedio (práctica relativamente segura)**:

    * `if "email" in person:` antes de acceder.  
      - Funciona, pero usa dos líneas para algo que se puede hacer en una.

    * `person.get("email")`
      - Accedes pero recibes un None, no verás nada


3. **Nivel experto (óptimo)**

    * `person.get("email", 0)`
      - **En una sola linea, sin crash, con un valor por defecto como "salvaguarda"**


---

> **Un Apunte sobre nulos necesarios:**  

> Python, desde 3.15, incluye un sistema de centinelas, para poder asignar valores textuales, numéricos o de cualquier tipo a lo que tradicionalmente sería un null `None`, sin modificar el comportamiento base de `None`.  
>   
> Y, esto, a la hora de trabajar con tablas, donde tus celdas null deben ser así en vez de valor `0`, para no destrozar posteriores cálculos y métricas (como medias, medianas, desviación, etc), **es fundamental**.
