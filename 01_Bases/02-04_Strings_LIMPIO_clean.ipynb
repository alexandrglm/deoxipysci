### CADENAS (`String`)

Una cadena,string, `str`, es el tipo de dato para almacenar **texto** (caracteres, paralabras, frases).  

Para definir un String, el texto va "entrecomillado".  
- Usar `"comillas dobles"` o `'comillas simples'` es cuestión de preferencia, aunque internamente Python suele trabajar con las simples.  
- Lo que importa es ser coherente con las comillas usadas en todo caso (o un `SyntaxError` te saludará).  


***

### PROPIEDADES `type` Y `length`

De las propiedades del objeto (tipo, longitud), a sus métodos `type(str)` y `len(str).  

*   `type(str)` devuelve el tipo de la variable `str`.  
*   `len(str)` devuelve su longitud, incluye todos los caracteres, espacios, cualquier cosa que lleve la cadena. **NO ES UN VALOR ZERO-BASED, sino literal**.  

***



### OPERACODRES CON STRINGS

Los operadores algebraicos funcionan con Strings, pero **cambian el comportamiento**.  



*   `+` - Suma, **hace concatenación** de Strings:
    - `"Hello " + "World" = "Hello World"`
*   `*` - Producto, **hace replicación**:
    - `"Hello " * 5 = "Hello Hello Hello Hello Hello"`


| Operador  | Con números (`Int` / `Float`) | Con cadenas (`String`)  |
| -- | -- | -- |
| `SUMA (+)`       | Suma: `5 + 5 = 10`            | **Concatena**: `"A" + "B" = "AB"` |
| `MULTIPLICACION (*)` | Multiplicac: `5 * 3 = 15`  | **Replica**: `"A" * 3 = "AAA"`    |


### PUNTERO DE MEMORIA CON OPERACIONES DE STRINGS

**Ejecutar una operación con Strings NO modifica la variable**.

```python
hello = "Hello "
world = "World!"

print(hello + world)
print(World * 3)
print( (hello + world) * 2 )

print(hello)
print(world)
```

Para guardar el resultado, y modificar el valor del objeto en memoria ralmente, se debe **reasignar explícitamente**:  

```python
hello = hello + world

print(hello)
```


### INTERPOLACION DE STRINGS (`f-strings`)

Aunque en la práctica final no se trabaja constantemente "con prints en consola", habrá multitud de casos en los que necesites crear una cadena que "incruste" variables (debug, reportes, logs,), lo que permite la interpolación de cadenas.  

*   Siempre precede la cadena con `f`
*   Las variables se escriben con llaves `{ }`
*   Sirve tanto con Strings simpes `" "`, como para heredocs interpolados `""" """`

```python
expediente = "0123/AB"
nombre = "Pepe"
apellido = "Perez"

print(f"El expediente figura a nombre de {apellido}, {nombre}")

print(f"""
- EXPEDIENTE: {expediente}
- TITULAR:    {apelliudo}, {nombre}
""")
```
  

- `f-string` **evalúa lo que hay dentro de las llaves como código Python, así que la variable debe existir con ese nombre exacto, o la expresión que le indiques debe ser sintácticamente correcta.**
