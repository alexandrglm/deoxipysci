### DATA STRUCTURES 1

#### ¿Qué es una estructura de datos?  
---

Hasta ahora, cada variable que hemos visto solo guardaba un tipo de dato o valor.  
  - **Una estructura de datos es un contenedor que agrupa distintos valores, bajo un mismo nombre**.  
  - La primera y más básica en Python es la **Lista `[valor1, valor2, ... ]`.  

***


### LISTAS `[ ]`

*   Se definen por corchetes (brackets), `[ ]`
*   Cada elemento va separado por comas,  `[valor_1, valor_2, ...]`

- Sin brackets no hay lista
- Sin comas que separen valores no hay lista
- **Una lista puede estar vacía `[]`
- Una lista **puede contener cualquier tipo ed dato, y puede mezclar varios tipos**


***



### INDICES:  Acceso por posición

Las listas **mantienen el orden en el que se le insertan los elementos**:  

* Cada elemento tiene una posición, llamada **index**  
* Python es **zero-based**, por lo que los index se ven influenciados por ello (Posición 1 es Index 0, Posicion 2 es Index 1, ...)  

Para acceder a un elemento, se escribe el nombre de la lista y, **entre brackets `[]` la posición, el index:  

```python
lista = [1, 2, 3, 4]

print(lista[0])
print(lista[1])
print(lista[2])
print(lista[3])
```
- Si se le pide un **index que no existe, te saludará un `IndexError`**.  

***



### MUTABILIDAD DE LISTAS:    Reasignar un elemento

**Las listas son mutables**, su contenido se puede modificar una vez creadas.  

- Es posible acceder a un elemento y reasignarle un nuevo valor:
```python
lista = [1, 2, 3, 4]

lista[0] = 100

print(lista) # [100, 2, 3, 4]
```
- Esto es lo que diferencia a una lista de un String (que es immutable, no puedes modificar caracteres unitarios sino la variable completa), o las tuplas (que veremos más adelante, y también son immutables).  
- Es la **característica principal, su mutabilidad**.  


***

### PROPIEDADES de LISTAS: Tipo, longitud

De la propiedad (`type`, `length`), al método (`type()`, len()`):  

- type

Por lo que, en una lista de $n$ elementos:
1. Los index válidos van de $0$ a $n - 1$
