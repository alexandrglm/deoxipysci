### `.append()`:    Añadir al final

Para añadir un elemento al final de una lista, se usa el método de estructuras de datos (de listas) `variable_lista.append(valor_que_se_añade)`:  

```python
lista = [1, 2, 3]

lista.append(4)

print(lista) # [1, 2, 3, 4]
```


##### ⛔ **ATENCION con `append`**

- `append()` **modifica la lista directamente**
    - **NUNCA** asignes a variable con `append` (`lista = lista.append(5)`) ni sobreescribas con `None`
- **Añade solo un elemento cada vez**, si pasas otra lista como argumento, anidará lista dentro de otra ( `[1, 2, 3, 4. [5, 6,] ]` )  

***


### INDEX NEGATIVOS

Además de acceder con valores positivos, las listas permiten acceder con valores negativos, **pero su comportamiento cambia, empezando por el último elemento de la lista**:

- `-1`, última posición
- `-2`, penúltima posición
- ...

```python
lista = [1, 2, 3, 4]

print(lista[-1]) # 4
print(lista[-2]) # 3
print(lista[-3]) # 2
print(lista[-4]) # 1
```

> Por ende, no existe un "negativo cero".  

***

### RANGOS:    Slicing básico con `[ : ]`

El slicing permite extraer un subconjunto de la lista usando rangos, siempre zero-based, con la sintáxis  `[inicio rango : fin rango]`, pero con unas características determinadas sobre si se incluyen o no:

- **El inicio SIEMPRE se incluye en el resultado**
- **El fin NO se incluye, se excluye del resultado**

```python
lista = [1, 2, 3, 4, 5, 6]

print(lista[0:4]) # 1, 2, 3, 4
print(lista[2:3]) # 3
```

---

#### SLICING OMITIENDO UN EXTREMO

También puedes **omitir uno de los extremos**:
  - `[2:]` - *Desde la posición 2 hasta el final*
  - `[:2]` - **Desde el inició hasta la posición 2* ( ⛔ **ESTE TAMBIEN ES UN CASO EXCLUYENTE**)

```python
lista = [1, 2, 3, 4, 5, 6]

print(lista[2:]) # 3, 4, 5, 6
print(lista[:2]) # 1, 2
```

---

#### SLICING NEGATIVOS `[-n:]`
Además, **funciona perfectamente con índices negativos:**  

  - `lista[-2:]`, **los dos últimos**
  - `lista[:-2]`, **desde el principio hasta los dos últimos  ( ⛔ **ESTE TAMBIEN ES UN CASO EXCLUYENTE**)

```python
lista = [1, 2, 3, 4, 5, 6]


print(lista[-2:]) # 5, 6
print(lista[:2-]) # 1, 2, 3, 4
```

---

#### SLICING EN MEMORIA

El slicing **crea una lista nueva, no modifica la original**:  

  - Esto es distinto de la asignación por índice (`my_list[1] = 99`), que sí modifica la posición.



***



### RANGOS CON PASO `[::n]`

EL tercer argumento o valor del slicing es el paso (step), cuya sintáxis total sería `[inicio:fin:paso]`:  
  * **De cuánto en cuánto queremos mostrar el slice**
    - Paso 1, por defecto, de uno en uno
    - Paso 2, de dos en dos,
    - Paso $n$, de $n$ en $n$
    - **Paso negativo ($-n$), en DIRECCION INVERSA**
  * El términon fin ⛔ **SIGUE SIENDO UN CASO EXCLUYENTE** y, este caso, debemos incluir EXPLICITAMENTE un $n + 1$ al valor de fin si realmente queremos incluirlo en nuestro slice.
  * ⛔ El término paso, al usar inversas en rangos, también debería usar signo negativo (`lista[-1::-2] = [6, 4, 2]` vs `lista[-1::2] = [6]`)


```python
lista = [1, 2, 3, 4, 5, 6]


print(lista[::2]) # 1, 3, 5
print(lista[::-1]) # 6, 5, 4, 3, 2, 1

print(lista[0:5:1]) # 1, 2, 3, 4, 5
print(lista[0:6:1]) # EXPLICITO PARA INCLUIR el SEXOT valor al término FIN, y, devuelve 1, 2, 3, 4, 5, 6
print(lista[0:5:2]) # 1, 3, 5

print(lista[-1::-2]) # [6, 4, 2]
print(lista[-1::2]) # [6]

```

  
-  Como curiosidad, `[::-1]` es la forma estándar para invertir una lista (en modo vista, sin modificarla).  

***



### PUNTEROS CON LISTAS

**Asignar una lista a otra variable NO crea una copia a una nueva variable:**  
  * Se crea un puntero a la misma dirección de memoria  
  * Todos los cambios que se hagan en una variable se aplican a la otra  
  * *¿Por qué?* Un dataset puede ocupar gigas y para cada visualización de un dato o slice **no es rentable ** qu se duplicase en memoria la lista


```python
a = [1, 2, 3]

b = a

print(a)
print(b)

b[0] = 100

print(b)
print(a)
```

Para **obtener una copia independiente, hay que hacerlo explícitamente**:  

* Con slicing completo, `b = a[:]`  
* Usando función `list()`, `b = list(a)`  
* Usando método `.copy()`, `b = a.copy()`


```python
a = [1, 2, 3]

b = a
c = a[:]
d = list(a)
e = a.copy()

a[0] = 100

print(a)
print(b)
print(c)
print(d)
print(e)

```

> **Está es la diferencia brutal entre mantener las variables y las operaciones o perderlas por el camino**

***

### OPERADORES y MÉTODOS sobre LISTAS

* `del my_list[i]`
  - Elimina el elemento en la posición `i`.  
  - No devuelve nada, modifica la lista en su sitio.  

* `list1 + list2`  
  - **CONCATENA** dos listas **CREANDO UNA NUEVA**.  
  - Mismo comportamiento que `+` pero con Strings.  

* `.extend(otra_lista)`
  - Extiende la lista actual con los elementos de la otra.
  - **Modifica en su sitio (NO CREA UNA NUEVA)**.  

* `.index(valor)`
  - Devuelve el **index** de la primera aparición de `valor` en la lista.  
  - Si no existe, te saluda un `ValueError`.  

* `.remove(valor)`
  - Elimina la primera aparición de `valor` (**por valor, no por index**).  
  - - **Modifica en su sitio (NO CREA UNA NUEVA)**.  

* `.insert(i, valor)`
  - Inserta `valor` en la posición `i`, **desplazando el resto**.
  - **Modifica en su sitio (NO CREA UNA NUEVA)**.  

---
Se repite en cada método si modifica (Vista) o crea una nueva variable, para tener muy claras las diferencias:  

- **`+` crea una lista nueva** (no toca las originales).
- **`.extend()` modifica la lista existente** (no crea nada nuevo).


---

```python
lista = [10, 20, 30]

lista.extend([40, 50])   # [10, 20, 30, 40, 50]

print(lista.index(30))   # 2

lista.remove(20)         # [10, 30, 40, 50]

lista.insert(1, 15)      # [10, 15, 30, 40, 50]

del lista[0]             # [15, 30, 40, 50]
```



#### FUNCIONES vs MËTODOS

Como curiosidad, y para hablar con propiedad respecto a métodos de cada tipo y funciones:  

- `print()`, `len()`, `type()` son **funciones** legacy de Python, se llaman por su nombre.

- `.append()`, `.extend()`, `.index()`, `.remove()`, `.insert()` son **métodos** para `List`:  
  - Pertenecen a la variable.  
  - Se invocan con punto.  
  - Por consecuencia, solo existen sobre este tipo específico, listas.
