# 02-02:  Numbers, and Mathematical Operations

## Números en Python

Los dos tipos numéricos fundamenteales son:

1.   Enteros - Integers - `int`:  
    *   Numeros enteros, positivos o no, **sin decimales**  
    * `1`, `-1`, `238786763`, ...  


2.   De coma flotante - Floats - `float`:   
    *   Números, positivos o no, que **incluyen decimales**  
    * `3.14`, `-12.223`, `0.000007`, ...  

En Python, **los decimales siempre se expresan con PUNTOS (`.`), nunca con coma (`,`)**.    

***



## INFERENCIA DE TIPOS

Cuando se combinan distintos tipos en alguna operación, como sumar un `int` con un `float`, **siempre se obtiene un `float` en respuesta**.  
  
Es decir, al mezclar tipos, Python decide cuál es el tipo más adecuado para responder, siendo los floats los más recurrentes.  

- `int` + `int` = `int`
- `int` + `float` (o la inversa) = `float`
- `int` / `int` con resto `0` = `float`

Se infiere el tipo de esta forma **para no perder precisión numérica**.  
  
  - El caso de las divisiones, que veremos en el siguiente punto, es particular y, siempre, devuelve un float, aunque sea entre enteros sin resto (10/2 = 5.0)


***


## OPERACIONES MATEMÁTICAS BÁSICAS

Python incluye los operadores aritméticos estándar, y alguno de más.


| Operador | Nombre | Ejemplo | Resultado |
|----------|--------|---------|-----------|
| `+` | Suma | `7 + 3` | `10` |
| `-` | Resta | `7 - 3` | `4` |
| `*` | Multiplicación | `7 * 3` | `21` |
| `/` | División | `7 / 3` | `2.3333…` |
| `//` | División entera | `7 // 3` | `2` |
| `%` | Módulo (resto) | `7 % 3` | `1` |
| `**` | Potencia | `7 ** 3` | `343` |

- Suma, resta y multiplicación operan como habitual.

- Para división, añadimos el concepto del "Resto de una división" con el operador modulus (Módulo), '%`:  
    * A grandes rasgos, es muy ítul para encontrar enteros, por ejemplo para operar sobre números primos o no, etc.

- El operador de división entera, `//` solo devuelve una operación **sin decimales**:
    * **No redondea de ninguna forma**
    * A diferencia de una división estándar, **siempre devuelve un entero**, `int`


***

### Python como calculadora I:   PRIORIDAD DE OPERACIONES (PEMDAS)

Python no evalúa una expresión de izquierda a derecha sin más, utiliza un orden de prioridad, el estándar asumido en matemáticas.  

- **DE MAYOR A MENOR PRIORIDAD**:

1. Potencias
2. Multiplicaciones y Divisiones (y restos)
3. Sumas y Restas

##### **PEMDAS**

Aforismo en álgebra que define perfectamente y a través de un acrónimo el orden exacto a seguir:  

> "Please, excuse my dear auntie Sophie!"

- **P**  - Paréntesis `()` (Parenthsis)
- **E**  - Exponentes (Potencias `**`, también raíces)
- **M**  - Multiplicaciones `*`
- **D**  - Divisiones `/` (y restos, `%`)
- **A**  - Sumas `+` (Additions)
- **S**  - Restas `-` (Substracions)

***

### Python como calculadora 2:   Divisiones siempre dan `float`

Hay dos divisiones en Python, pero se comportan de forma diferente:  

| Operador | Nombre | Devuelve ... | Ejemplo |
|---|---|---|---|
| `/`   | **División Real** | `float` (**SIEMPRE**) | `10 / 2 = 5.0`, `10 / 6 = 1.6666666666666667`, `10 / 3 = 3.3333333333333335`
   |
| `//`  | **División Entera** | Descarta decimales, no redondea (trunca hacia abajo), `int` | `10 / 2 = 5`, `10 // 6 = 1`, `10 // 3 = 3`|
  
- **Contraintuitivamente, la división entera con negativos, sigue truncando hacia abajo en vez de hacia arriba**:
    * `-7 // 2 = -4`
    * Por qué `-4` en vez de `-3`?
    * Con la División Entera lo que estamos haciendo es "¿Cuantas veces cabe el dividendo entre el divisor?"
    * Entonces, ¿Cuántas veces cabe 2 en -7 sin pasarse?, la respuesta es -4, no -3.

### Python como Calculadora 3: Precición real de los `float`'s

Como apunte, los `float` **no son infinitamente precisos**, ni en Python, ni en otros lenguajes de programación.  

La consecuencia más conocida:  


````python
print("1. 0.1 sumado a 0.2 son: ", 0.1 + 0.2)
falso = (0.1 + 0.2 == 0.3)
print(f"2. Si la suma de 0.1 y 0.2 es 0.3... Cómo es posible que su comparación directa sea: {falso} ?")
````

**Aunque se detalla más adelante, operadores lógicos, de comparación, está bien tener en cuenta desde este momento donde se definen los números y propiedades, el hecho de la imprecisión de `float`s**:

Para hacer comparaciones, nunca se usan comparaciones directas `==`.  

Bien usando métodos propios de Python, o librerías, se compara buscando la similitud o cercanía "relativa", por ejemplo:  

- Redondeando con `.round()` (aunque esto hace perder la precisión)
- Usando el módulo `Decimal`, para operaciones con precisión financiera
- En `NumPy` usando sus métodos `.isclose()` que incluso permite trabajar con magnitudes muy altas (o bajas) y definir su tolerancia, o para matrices con métodos más avanzados como `.linalg.matrix_rank()`

### Operadores de asignación (Assignment Operators)

Son **atajos de código** para actualizar el valor de una variable combinando una operación aritmética con una asignación, todo en una sola línea.

Aunque son infinitamente útiles como "contadores" o valores que se incrementan "por cada operación dada" (en los `for ... in`), y habitualmente usamos el operador de asignación de suma `+=`, realmente se puede usar como atajo cualquier símbolo aritmético básico como operador de asignación.  
  

| Operador | Equivalencia | Ejemplo (siendo `x = 10`) | `x` queda en |
|-------|-------------|----------------------------------|--------------|
| `x += 3` | `x = x + 3` | `10 + 3` | `13` |
| `x -= 3` | `x = x - 3` | `10 - 3` | `7` |
| `x *= 3` | `x = x * 3` | `10 * 3` | `30` |
| `x /= 3` | `x = x / 3` | `10 / 3` | `3.333…` |
| `x //= 3` | `x = x // 3` | `10 // 3` | `3` |
| `x %= 3` | `x = x % 3` | `10 % 3` | `1` |
| `x **= 3` | `x = x ** 3` | `10 ** 3` | `1000` |


**Se leen como** *"a `x`, asígnale `x` [operación] 3"*. Es decir, `x += 3` significa *"a `x`, asígnale `x` más 3"*.

- **OPERAR VS ASIGNAR**  
  Cuando usas un operador de asignación, **automáticamente estás guardando el valor de dicha variable incrementado en la forma en que lo asignes**.  

- **EL TIPO TAMBIEN SE INFIERE**  
  De la misma forma que con cualquier otra operación, el tipo también se infiere, y una variable puede cambiar después de asignar operación.  



### ¿Por qué existen estos atajos?

- **Legibilidad**: `total += item_price` se lee mejor que `total = total + item_price`.
- **Menos errores**: no repites el nombre de la variable a ambos lados del `=`, así que no puedes equivocarte escribiéndolo distinto.
- **Idiomático**: es el estilo estándar en Python y en cualquier lenguaje que los soporte (C, Java, JS, etc.).

> 💡 **En IA y Data Science** los verás constantemente en acumuladores: sumar pérdidas en un bucle de entrenamiento (`total_loss += loss`), contar aciertos (`correct += 1`), acumular gradientes, etc.
