### LOGICA BOOLEANA

El tipo booleano existe para trabajar con lógica binaria:
*   Comparar, filtrar, controlar y dirigir el flujo del código  
*   Solo hay dos valores posibles: `True`(numericamente,`1`), o `False` (numéricamente, `0`)  
* Sobre todo esto se sustentan los **operadores lógicos** más básicos:
  1.  **AND**
  2.  **OR**
  3.  **NOT**

#### AND - "Y" lógico

`and` **es estricto** (tal condición AND tal otra condición).  
Todos las condiciones dadas deben ser `True` de lo contrario, emitirá `False`

*   `True` and `True` = `True`
*   `False` and `False` = `False`
*   `False` and `True` = `False` (y viceversa)

#### OR - "O" lógico

`or` es más fléxible. Con que una sola de las condiciones se cumplan, en suficiente para emitir `True`.  

*   `False` and `True` = `True` (y viceversa)
*   `True` or `True` = `True`
*   `False` or `False` = `False`



#### NOT - Negación
`not` invierte la condición.

*   `not True` = `False`
*   `not False` = `True`




***

### OPERADORES DE COMPARACION

Son la base del filtrado de datos y el control del flujo de un programa.  

Los distintos operadores **siempre van a devolver un booleano**.

| Operador | Significa | Ejemplo | Resultado |
|-|-|-|-|
| `==` | **igual a** | `5 == 5` | `True` |
| `!=` | **distinto de** | `5 != 3` | `True` |
| `<` | **menor que** | `5 < 3` | `False` |
| `>` | **mayor que** | `5 > 3` | `True` |
| `<=` | **menor o igual que** | `5 <= 5` | `True` |
| `>=` | **mayor o igual que** | `5 >= 6` | `False` |


- `=` asignaba un valor, `==` compara valores.  
- Se pueden encadenar comparaciones, `0 < x <= 200`, que Python lo evalúa "sintácticamente" como `(0 < x) and (x <= 200)`.  
- Funciona con cualquier tipo numérico, `int` `float` (e, incluso con `String` en ciertos casos, comparación de caracteres o equivalencias de dos cadenas, aunque no es el propósito).  



### NUMEROS COMO BOOLEANOS (Truthy / Falsy)

**Cualquier valor numérico puede evaluarse como booleano**:  

- `0` **siempre es `False`**. Ningún otro valor puede ser cero.  
- Cualquier otro número, positivo o negativo, siempre es `True`


Pero, siempre dentro de un contexto lógico, es decir, que `0` "se compornte" como `False` no significa que en una comparación directa `==` sea `False`.  

***

### TABLA DE LA VERDAD AND - OR

| Valor 1 | Operador | Valor 2 | = Resultado Final |
| - | - | - | - |
| `True` | `and` | `True` | `True` |
| `True` | `and` | `False` | `False` |
| `False` | `and` | `False` | `False` |
| `True` | `or` | `False` | `True` |
| `False` | `or` | `False` | `False` |
