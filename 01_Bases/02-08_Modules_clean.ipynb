## MODULOS

Un módulo es un script de Python escrito por alguien que contiene funciones y utilidades listas para usar:  

  * En vez de programar uno mismo la lógica, la importas y puedes usar todas las funciones directamente.  



> No hay que reinventar la rueda. Si ya existe una función optimizada para álgebra lineal, no hay que reinventarla.  

Existen dos tipos de módulos:

1. **NATIVOS** (`math`, `time`, `random`, `sys`)  
    * Vienen con Python, son legacy.  

2. **EXTERNOS** (NumPy, Pandas, ...)  
    * Se instalan primero con `pip install`.  

***



#### Las tres formas de importar


1. **IMPORT COMPLETO**  `import modulo`
    * Carga el módulo entero en memoria.  
    * Se accede a sus funciones con notación de punto (`modulo.funcion()`)
```python
import math

print(math.sqrt(25))

```

2. **IMPORT ESPECÍFICO**  `from modulo import funcion`
    * Se carga solo la funcion que se necesita.  
    * Se invoca directamente, sin prefijos.  
```python
from math import sqrt

print(sqrt(25))

```

3. **IMPORT CON ALIAS**   `import modulo as alias`
    * Otorga un apodo corto al módulo.  
    * Es lo estándar (`import numpy as np`, `import pandas as pd`)  
```python
import math as m

print(m.sqrt(25))
```

***

#### `import` vs `from ... import`

| Forma | Ventaja | Inconveniente |
|-------|---------|---------------|
| `import math` | Tienes todo el módulo disponible | Hay que escribir el prefijo siempre |
| `from math import sqrt` | Llamada directa, más corta | Solo tienes lo que importaste |

  * Si vas a usar **muchas** funciones del módulo : `import module`  
  * Si solo **una o dos**:  `from module import function`.  

---

Errores típicos son:  

- **`ModuleNotFoundError`**: Python no encuentra el módulo.  
  * Causas: nombre mal escrito, o es externo y no está instalado.

- **`NameError` por prefijo faltante**: hiciste `import math` pero llamas `sqrt()` sin `math.`.  
  * Hay que usar el prefijo.

- **Mala práctica: importar a mitad del código**.  
  * Los `import` van siempre al principio del archivo.

- ⛔ **Nunca mezcles una forma con otra**
  * Si haces `import math`, no puedes llamar `sqrt(25)` directamente.  
  * Te saludará un `NameError`. Debes escribir `math.sqrt(25)`.
